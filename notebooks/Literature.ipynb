**Goal**: A defensible contribution is a controlled comparison of CNN and Transformer models under the same computational budget and evaluation protocol, including both segmentation quality and efficiency.

## Which architectures are common, and why?

The literature on medical image segmentation commonly uses encoder-decoder CNNs, residual networks, and increasingly Transformer-based or hybrid architectures. Each represents a different approach to extracting and combining visual information.

CNNs and Vision Transformers are two major approaches in medical image analysis. A [2024 systematic review](https://link.springer.com/article/10.1007/s10916-024-02105-8) of their use in medical imaging discusses their performance, computational efficiency, and the importance of pretraining for Transformers.


**CNNs: ResNet, DenseNet, EfficientNet**

Convolutions learn local patterns such as edges, texture, color transitions, and tissue boundaries. These features are useful for endoscopic images, where subtle local visual details can distinguish classes.

* ResNet: residual connections help optimize deeper networks.
* DenseNet: feature reuse through densely connected layers.
* EfficientNet: balances network depth, width, and input resolution to improve efficiency.

**Transformers: ViT and Swin Transformer**

Transformers divide images into patches and use attention to learn relationships between different image regions. This can help distinguish patterns that depend on broader context, rather than only nearby pixels.

* ViT: a straightforward global-attention baseline.
* Swin Transformer: hierarchical features and window-based attention, often a more scalable design for vision tasks.

**U-Net — CNN encoder-decoder**

*Local features and precise boundaries*

U-Net combines a contracting path that learns visual features with an expanding path that reconstructs a pixel-level mask. Skip connections preserve spatial detail, which is important for identifying small polyps and their boundaries. Its strong performance with relatively small medical datasets makes it an appropriate baseline.

**DeepLabV3+ — CNN with multiscale context**

*Context at different spatial scales*

DeepLabV3+ uses atrous convolutions and an Atrous Spatial Pyramid Pooling module to capture information at different scales. This is useful when polyps vary in size, shape, and appearance. Its decoder helps recover spatial detail lost during feature extraction.

**SegFormer-B0 — Transformer-based segmentation**

*Global context and hierarchical features*

SegFormer uses a hierarchical Transformer encoder and a lightweight MLP decoder. Attention mechanisms help model relationships between distant image regions, potentially helping distinguish polyps from surrounding tissue. The B0 variant is a relatively compact option for experiments with limited compute.
